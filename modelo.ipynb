```python
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder, OneHotEncoder
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
try:
    from imblearn.over_sampling import SMOTE
    _SMOTE_AVAILABLE = True
except Exception:
    _SMOTE_AVAILABLE = False
```

```python
def agrupar_classes(classe):
    if classe in ["Insufficient_Weight", "Normal_Weight", "Overweight_Level_I", "Overweight_Level_II"]:
        return "Não obeso"
    elif classe == "Obesity_Type_I":
        return "Obeso tipo 1"
    elif classe == "Obesity_Type_II":
        return "Obeso tipo 2"
    elif classe == "Obesity_Type_III":
        return "Obeso tipo 3"
    else:
        return classe

# Aplicar agrupamento
df_dados_ptbr["Classe_peso_corporal"] = df_dados_ptbr["Classe_peso_corporal"].apply(agrupar_classes)

print("Distribuição das classes agrupadas:")
print(df_dados_ptbr["Classe_peso_corporal"].value_counts(normalize=True) * 100)

# Mapear binários e ordinais básicos antes da pipeline (garante valores esperados)
df_dados_ptbr["Genero"] = df_dados_ptbr["Genero"].map({"F": 1, "M": 0})
ordem_consumo = {"no": 0, "Sometimes": 1, "Frequently": 2, "Always": 3}
df_dados_ptbr["Consumo_entre_refeicoes"] = df_dados_ptbr["Consumo_entre_refeicoes"].map(ordem_consumo)
df_dados_ptbr["Consumo_alcoolico"] = df_dados_ptbr["Consumo_alcoolico"].map(ordem_consumo)

# Transformer helpers (estilo do exemplo anexado)
from sklearn.base import BaseEstimator, TransformerMixin
```

```python
class GroupObesityClasses(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        X = X.copy()
        X['Classe_peso_corporal'] = X['Classe_peso_corporal'].apply(agrupar_classes)
        return X
```

```python
class BinaryOrdinalEncoder(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.genero_map = {"F": 1, "M": 0}
        self.ordem_consumo = {"no": 0, "Sometimes": 1, "Frequently": 2, "Always": 3}
    
    def fit(self, X, y=None):
        return self
    
    def transform(self, X):
        X = X.copy()
        if 'Genero' in X.columns:
            X['Genero'] = X['Genero'].map(self.genero_map)
        if 'Consumo_entre_refeicoes' in X.columns:
            X['Consumo_entre_refeicoes'] = X['Consumo_entre_refeicoes'].map(self.ordem_consumo)
        if 'Consumo_alcoolico' in X.columns:
            X['Consumo_alcoolico'] = X['Consumo_alcoolico'].map(self.ordem_consumo)
        return X
```

```python
class DropFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, features_to_drop=['Peso', 'Altura']):
        self.features_to_drop = features_to_drop
    def fit(self, X, y=None):
        return self
    def transform(self, X):
        X = X.copy()
        for f in self.features_to_drop:
            if f in X.columns:
                X.drop(f, axis=1, inplace=True)
        return X
```

```python
class MinMax(BaseEstimator, TransformerMixin):
    def __init__(self, numeric_cols=['Idade','Frequencia_consumo_vegetais','Numero_refeicoes_diarias',
                                      'Consumo_agua_diario','Frequencia_semanal_atividade_fisica','Tempo_diario_dispositivos_eletronicos']):
        self.numeric_cols = numeric_cols
        self.scaler = StandardScaler()
    def fit(self, X, y=None):
        self.scaler.fit(X[self.numeric_cols])
        return self
    def transform(self, X):
        X = X.copy()
        X[self.numeric_cols] = self.scaler.transform(X[self.numeric_cols])
        return X
```

```python
class OneHotEncodingNames(BaseEstimator, TransformerMixin):
    def __init__(self, onehot_cols=['Meio_transporte_principal']):
        self.onehot_cols = onehot_cols
        self.enc = None
        self.feature_names = None
    def fit(self, X, y=None):
        self.enc = OneHotEncoder(sparse=False, handle_unknown='ignore')
        self.enc.fit(X[self.onehot_cols])
        self.feature_names = self.enc.get_feature_names_out(self.onehot_cols)
        return self
    def transform(self, X):
        X = X.copy()
        oh = pd.DataFrame(self.enc.transform(X[self.onehot_cols]), columns=self.feature_names, index=X.index)
        X = pd.concat([X.drop(columns=self.onehot_cols), oh], axis=1)
        return X
```

```python
class OrdinalFeature(BaseEstimator, TransformerMixin):
    def __init__(self, mapping=None):
        self.mapping = mapping or {"Consumo_entre_refeicoes": ordem_consumo, "Consumo_alcoolico": ordem_consumo}
    def fit(self, X, y=None):
        return self
    def transform(self, X):
        X = X.copy()
        for col, mp in self.mapping.items():
            if col in X.columns:
                X[col] = X[col].map(mp)
        return X
```

```python
from sklearn.pipeline import Pipeline
```

```python
# Agrupar as classes de obesidade conforme definido




# Construir pipeline no estilo do exemplo (sem oversample embutido)
pipeline_preproc = Pipeline([
    ('group_classes', GroupObesityClasses()),
    ('bin_ordinal', BinaryOrdinalEncoder()),
    ('drop', DropFeatures()),
    ('ordinal', OrdinalFeature()),
    ('onehot', OneHotEncodingNames(onehot_cols=['Meio_transporte_principal'])),
    ('scale_numeric', MinMax())
])

# Aplicar pipeline de pré-processamento ao dataset completo
df_pre = pipeline_preproc.fit_transform(df_dados_ptbr)

# Preparar X e y (agora sem Peso/Altura)
X = df_pre.drop(columns=['Classe_peso_corporal'])
y = df_pre['Classe_peso_corporal']

print("\nFeatures usadas no modelo:")
print(X.columns.tolist())
print(f"\nShape de X: {X.shape}")
print(f"Shape de y: {y.shape}")
```

```python
# Imputar valores faltantes em X (numéricos: mediana, categóricos: moda)
from sklearn.impute import SimpleImputer

# Verificar colunas numéricas e categóricas
numeric_cols = X.select_dtypes(include=[np.number]).columns.tolist()
cat_cols = X.select_dtypes(exclude=[np.number]).columns.tolist()

print(f"Colunas numéricas ({len(numeric_cols)}): {numeric_cols}")
print(f"Colunas categóricas ({len(cat_cols)}): {cat_cols}")

# Imputation de colunas numéricas
if numeric_cols:
    imp_num = SimpleImputer(strategy='median')
    arr_num = imp_num.fit_transform(X[numeric_cols])
    # Usar o tamanho real do array retornado
    numeric_cols_actual = numeric_cols[:arr_num.shape[1]]
    X_num_imp = pd.DataFrame(arr_num, columns=numeric_cols_actual, index=X.index)
    X[numeric_cols_actual] = X_num_imp

# Imputation de colunas categóricas
if cat_cols:
    imp_cat = SimpleImputer(strategy='most_frequent')
    arr_cat = imp_cat.fit_transform(X[cat_cols])
    # Usar o tamanho real do array retornado
    cat_cols_actual = cat_cols[:arr_cat.shape[1]]
    X_cat_imp = pd.DataFrame(arr_cat, columns=cat_cols_actual, index=X.index)
    X[cat_cols_actual] = X_cat_imp

# Verificar se ainda há NaNs
print('\nMissing por coluna após imputação:')
missing = X.isnull().sum()[X.isnull().sum() > 0]
if len(missing) > 0:
    print(missing)
else:
    print("Nenhum valor faltante!")

# Codificar y como labels numéricos para o treinamento
le = LabelEncoder()
y_enc = le.fit_transform(y)

# Dividir treino/teste
X_train, X_test, y_train, y_test = train_test_split(X, y_enc, test_size=0.2, random_state=42, stratify=y_enc)

print(f"\nTamanho treino: {X_train.shape}")
print(f"Tamanho teste: {X_test.shape}")

# Aplicar SMOTE apenas no conjunto de treino (fallback para ambientes sem imbalanced-learn)
if _SMOTE_AVAILABLE:
    sm = SMOTE(random_state=42)
    X_train_res, y_train_res = sm.fit_resample(X_train, y_train)
else:
    # fallback: upsample minority classes with sklearn.utils.resample
    from sklearn.utils import resample
    X_train_df = X_train.copy()
    y_train_ser = pd.Series(y_train, index=X_train_df.index, name='target')
    Xy = pd.concat([X_train_df, y_train_ser], axis=1)
    max_count = Xy['target'].value_counts().max()
    groups = []
    for cls, grp in Xy.groupby('target'):
        if len(grp) < max_count:
            grp_up = resample(grp, replace=True, n_samples=max_count, random_state=42)
            groups.append(grp_up)
        else:
            groups.append(grp)
    Xy_res = pd.concat(groups).sample(frac=1, random_state=42).reset_index(drop=True)
    y_train_res = Xy_res['target'].values
    X_train_res = Xy_res.drop(columns=['target'])

# Escalar (já usamos StandardScaler internamente no MinMax, mas padronizamos novamente para o classificador)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_res)
X_test_scaled = scaler.transform(X_test)

# Treinar RandomForest
modelo = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42, n_jobs=-1)
modelo.fit(X_train_scaled, y_train_res)

print("\nModelo treinado!")

# Avaliar
y_pred_test = modelo.predict(X_test_scaled)
print(f"\nAcurácia teste: {accuracy_score(y_test, y_pred_test):.4f}")
print("\n=== RELATÓRIO TESTE ===")
print(classification_report(y_test, y_pred_test, target_names=le.classes_))
print("\nMatriz de confusão:")
print(confusion_matrix(y_test, y_pred_test))

# Importância das features (top 10)
import matplotlib.pyplot as plt
importancias = modelo.feature_importances_
features = X.columns.to_list()
indices_ordenados = np.argsort(importancias)[::-1][:10]
plt.figure(figsize=(10,6))
plt.title("Top 10 Features Mais Importantes")
plt.bar(range(len(indices_ordenados)), importancias[indices_ordenados])
plt.xticks(range(len(indices_ordenados)), [features[i] for i in indices_ordenados], rotation=45, ha='right')
plt.xlabel("Feature")
plt.ylabel("Importância")
plt.tight_layout()
plt.show()

# Exemplo: previsão em um novo paciente (primeira linha do conjunto de teste)
novo_paciente = X_test.iloc[0:1]
novo_paciente_scaled = scaler.transform(novo_paciente)
pred = modelo.predict(novo_paciente_scaled)[0]
probs = modelo.predict_proba(novo_paciente_scaled)[0]
print(f"\nPredição: {le.inverse_transform([pred])[0]}")
print(f"Confiança: {max(probs):.2%}")
plt.show()
```
