### Verificação: Obesidade Grau 3 vs Consumo de Vegetais
Existe uma hipótese de que todos os dados de Obesidade Grau 3 possuem consumo frequente de vegetais (FCVC).

```python
## Análise Estatística Percentual por Categoria de Peso
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

features_para_analisar = [
    'Genero', 'Historico_familiar_sobrepeso', 'Consumo_alimentos_hipercaloricos', 
    'Frequencia_consumo_vegetais', 'Numero_refeicoes_diarias', 'Consumo_entre_refeicoes', 
    'Fumante', 'Consumo_agua_diario', 'Monitora_ingestao_calorica', 
    'Frequencia_semanal_atividade_fisica', 'Tempo_diario_dispositivos_eletronicos', 
    'Consumo_alcoolico', 'Meio_transporte_principal'
]

ordem_classes = [
    'Insufficient_Weight', 'Normal_Weight', 'Overweight_Level_I', 
    'Overweight_Level_II', 'Obesity_Type_I', 'Obesity_Type_II', 'Obesity_Type_III'
]

for feature in features_para_analisar:
    # Calculando a tabela de contingência (frequência absoluta)
    df_cont = pd.crosstab(df_dados_ptbr['Classe_peso_corporal'], df_dados_ptbr[feature])
    
    # Normalizando por linha para obter a porcentagem dentro de cada classe de peso
    df_perc = df_cont.div(df_cont.sum(axis=1), axis=0) * 100
    
    # Reorganizando para plotagem
    df_plot = df_perc.reset_index().melt(id_vars='Classe_peso_corporal')
    
    plt.figure(figsize=(14, 7))
    ax = sns.barplot(data=df_plot, x='Classe_peso_corporal', y='value', hue=feature, order=ordem_classes, palette='viridis')
    
    plt.title(f'Distribuição Percentual de {feature} por Classe de Obesidade')
    plt.ylabel('Porcentagem (%)')
    plt.xlabel('Classe de Peso')
    plt.xticks(rotation=45)
    plt.legend(title=feature, bbox_to_anchor=(1.05, 1), loc=2)
    plt.ylim(0, 110)
    
    # Adicionando rótulos de porcentagem nas barras
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f'{p.get_height():.1f}%', 
                        (p.get_x() + p.get_width() / 2., p.get_height()), 
                        ha = 'center', va = 'center', 
                        xytext = (0, 9), 
                        textcoords = 'offset points', 
                        fontsize=8, rotation=90)
    
    plt.tight_layout()
    plt.show()
    print('-'*100)
```

**Preparando para o modelo**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Modelos para avaliação
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

try:
    from imblearn.over_sampling import SMOTE
    _SMOTE_AVAILABLE = True
except ImportError:
    _SMOTE_AVAILABLE = False
```

### Definição dos Transformadores

```python
class GroupObesityClasses(BaseEstimator, TransformerMixin):
    def fit(self, df, y=None): return self
    def transform(self, df):
        return df.copy()
        def agrupar(classe):
            if not isinstance(classe, str): return classe
            if classe in ["Insufficient_Weight", "Normal_Weight"]:
                return "Peso Normal"
            elif classe in ["Overweight_Level_I", "Overweight_Level_II"]:
                return "Sobrepeso"
            elif classe == "Obesity_Type_I": return "Obeso tipo 1"
            elif classe == "Obesity_Type_II": return "Obeso tipo 2"
            elif classe == "Obesity_Type_III": return "Obeso tipo 3"
            return classe
            if classe in ["Insufficient_Weight", "Normal_Weight", "Overweight_Level_I", "Overweight_Level_II"]:
                return "Não obeso"
            elif classe == "Obesity_Type_I": return "Obeso tipo 1"
            elif classe == "Obesity_Type_II": return "Obeso tipo 2"
            elif classe == "Obesity_Type_III": return "Obeso tipo 3"
            return classe
        df['Classe_peso_corporal'] = df['Classe_peso_corporal'].apply(agrupar)
        return df

class DropFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, drop=['Escala_Obesidade', 'Perfil_Risco', 'Genero_num', 'Historico_num', 'Comida_Calorica_num', 'Lanches_num', 'Fumante_num', 'Monitora_Caloria_num']): self.drop = drop
    def fit(self, df, y=None): return self
    def transform(self, df):
        df = df.copy()
        cols = [f for f in self.drop if f in df.columns]
        return df.drop(cols, axis=1)

class OneHotEncodingNames(BaseEstimator, TransformerMixin):
    def __init__(self, cols=['Meio_transporte_principal']): 
        self.cols = cols
        self.enc = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
    def fit(self, df, y=None): 
        self.enc.fit(df[self.cols])
        return self
    def transform(self, df):
        df = df.copy()
        oh = pd.DataFrame(self.enc.transform(df[self.cols]), columns=self.enc.get_feature_names_out(self.cols), index=df.index)
        return pd.concat([oh, df.drop(self.cols, axis=1)], axis=1)

class OrdinalFeature(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.map_bin = {"F": 1, "M": 0}
        self.map_ord = {"no": 0, "Sometimes": 1, "Frequently": 2, "Always": 3}
    def fit(self, df, y=None): return self
    def transform(self, df):
        df = df.copy()
        if 'Genero' in df.columns: df['Genero'] = df['Genero'].map(self.map_bin)
        for c in ['Consumo_entre_refeicoes', 'Consumo_alcoolico']:
            if c in df.columns: df[c] = df[c].map(self.map_ord)
        return df

class MinMax(BaseEstimator, TransformerMixin):
    def __init__(self, cols=None):
        self.cols = cols or ['Altura', 'Peso', 'Altura', 'Idade', 'Frequencia_consumo_vegetais', 'Numero_refeicoes_diarias', 'Consumo_agua_diario', 'Frequencia_semanal_atividade_fisica', 'Tempo_diario_dispositivos_eletronicos']
        self.scaler = MinMaxScaler()
    def fit(self, df, y=None): 
        self.scaler.fit(df[self.cols])
        return self
    def transform(self, df):
        df = df.copy()
        df[self.cols] = self.scaler.transform(df[self.cols])
        return df

class Oversample(BaseEstimator, TransformerMixin):
    def __init__(self, seed): self.seed = seed
    def fit(self, df, y=None): return self
    def transform(self, df):
        if 'Classe_peso_corporal' in df.columns and _SMOTE_AVAILABLE:
            sm = SMOTE(sampling_strategy='not majority', random_state=self.seed)
            X_bal, y_bal = sm.fit_resample(df.drop('Classe_peso_corporal', axis=1), df['Classe_peso_corporal'])
            return pd.concat([pd.DataFrame(X_bal), pd.DataFrame(y_bal, columns=['Classe_peso_corporal'])], axis=1)
        return df
```

#**Avaliação de Estabilidade Multi-Seed**

```python
SEEDS = [10, 4232, 9999]
model_list = {
    "Logistic Regression": LogisticRegression(max_iter=2000),
    "Decision Tree": DecisionTreeClassifier(max_depth=15, class_weight='balanced'),
    "Random Forest": RandomForestClassifier(n_estimators=500, max_depth=25, class_weight='balanced', random_state=MAIN_SEED),
    "Extra Trees": ExtraTreesClassifier(n_estimators=500, max_depth=30, class_weight='balanced', random_state=MAIN_SEED),
    "AdaBoost": AdaBoostClassifier(n_estimators=300),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=1000, learning_rate=0.05, max_depth=6, random_state=MAIN_SEED),
    "SVM": SVC(probability=True, class_weight='balanced')
}

final_results = pd.DataFrame(index=model_list.keys())

print("Iniciando bateria de testes...")
for s in SEEDS:
    acc_per_seed = []
    # Split
    train_df, test_df = train_test_split(df_dados_ptbr, test_size=0.2, random_state=s, stratify=df_dados_ptbr['Classe_peso_corporal'])
    
    # Pipeline
    pipeline = Pipeline([
        ('group', GroupObesityClasses()),
        ('drop', DropFeatures()),
        ('ordinal', OrdinalFeature()),
        ('onehot', OneHotEncodingNames()),
        ('scale', MinMax()),
        ('oversample', Oversample(seed=MAIN_SEED))
    ])
    
    df_tr = pipeline.fit_transform(train_df)
    val_pipe = Pipeline(pipeline.steps[:-1])
    df_te = val_pipe.transform(test_df)
    
    X_tr, y_tr = df_tr.drop('Classe_peso_corporal', axis=1), df_tr['Classe_peso_corporal']
    X_te, y_te = df_te.drop('Classe_peso_corporal', axis=1), df_te['Classe_peso_corporal']
    
    for name, model in model_list.items():
        model.set_params(random_state=s) if hasattr(model, 'random_state') else None
        model.fit(X_tr, y_tr)
        acc = accuracy_score(y_te, model.predict(X_te))
        acc_per_seed.append(acc)
    
    final_results[f'Seed_{s}'] = acc_per_seed

final_results['Média'] = final_results.mean(axis=1)
final_results = final_results.sort_values(by='Média', ascending=False)

print("\n=== RANKING DE ESTABILIDADE ===")
display(final_results)

melhor_nome = final_results.index[0]
print(f"\nO modelo campeão (mais estável) é: {melhor_nome}")
```

### Treinamento Final e Visualização

```python
# Treinamos o vencedor com o SEED principal (4232) para exportação
MAIN_SEED = 4232
train_df, test_df = train_test_split(df_dados_ptbr, test_size=0.2, random_state=MAIN_SEED, stratify=df_dados_ptbr['Classe_peso_corporal'])

final_pipeline = Pipeline([
    ('group', GroupObesityClasses()),
    ('drop', DropFeatures()),
    ('ordinal', OrdinalFeature()),
    ('onehot', OneHotEncodingNames()),
    ('scale', MinMax()),
        ('oversample', Oversample(seed=MAIN_SEED))
])

df_tr_final = final_pipeline.fit_transform(train_df)
val_pipeline = Pipeline(final_pipeline.steps[:-1])
df_te_final = val_pipeline.transform(test_df)

X_tr_final, y_tr_final = df_tr_final.drop('Classe_peso_corporal', axis=1), df_tr_final['Classe_peso_corporal']
X_te_final, y_te_final = df_te_final.drop('Classe_peso_corporal', axis=1), df_te_final['Classe_peso_corporal']

vencedor = model_list[melhor_nome]
vencedor.set_params(random_state=MAIN_SEED) if hasattr(vencedor, 'random_state') else None
vencedor.fit(X_tr_final, y_tr_final)

y_pred = vencedor.predict(X_te_final)
labels = ["Insufficient_Weight", "Normal_Weight", "Overweight_Level_I", "Overweight_Level_II", "Obesity_Type_I", "Obesity_Type_II", "Obesity_Type_III"]

print(f"\n=== RELATÓRIO FINAL: {melhor_nome} ===")
print(classification_report(y_te_final, y_pred, target_names=labels))

cm = confusion_matrix(y_te_final, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title(f'Matriz de Confusão Final ({melhor_nome})')
plt.show()
```

## Impacto Geral: Quais variáveis mais impulsionam o aumento da classe de peso?

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Criando uma escala numérica para a Classe de Peso (0 a 6)
ordem_escala = {
    'Insufficient_Weight': 0,
    'Normal_Weight': 1,
    'Overweight_Level_I': 2,
    'Overweight_Level_II': 3,
    'Obesity_Type_I': 4,
    'Obesity_Type_II': 5,
    'Obesity_Type_III': 6
}
df_analise = df_dados_ptbr.copy()
df_analise['Escala_Obesidade'] = df_analise['Classe_peso_corporal'].map(ordem_escala)

# 2. Função robusta para garantir que variáveis sejam numéricas antes da correlação
def para_numerico(df, col, mapeamento):
    if df[col].dtype == 'object':
        return df[col].map(mapeamento)
    return df[col]

df_analise['Genero_num'] = para_numerico(df_analise, 'Genero', {'F': 1, 'M': 0})
df_analise['Historico_num'] = para_numerico(df_analise, 'Historico_familiar_sobrepeso', {'yes': 1, 'no': 0, 1: 1, 0: 0})
df_analise['Comida_Calorica_num'] = para_numerico(df_analise, 'Consumo_alimentos_hipercaloricos', {'yes': 1, 'no': 0, 1: 1, 0: 0})
df_analise['Lanches_num'] = para_numerico(df_analise, 'Consumo_entre_refeicoes', {'no': 0, 'Sometimes': 1, 'Frequently': 2, 'Always': 3})
df_analise['Fumante_num'] = para_numerico(df_analise, 'Fumante', {'yes': 1, 'no': 0, 1: 1, 0: 0})
df_analise['Monitora_Caloria_num'] = para_numerico(df_analise, 'Monitora_ingestao_calorica', {'yes': 1, 'no': 0, 1: 1, 0: 0})

# 3. Calculando a Correlação com a Progressão de Peso
cols_impacto = [
    'Escala_Obesidade', 'Idade', 'Genero_num', 'Historico_num', 
    'Comida_Calorica_num', 'Lanches_num', 'Frequencia_consumo_vegetais', 
    'Numero_refeicoes_diarias', 'Consumo_agua_diario', 'Fumante_num', 'Monitora_Caloria_num',
    'Frequencia_semanal_atividade_fisica', 'Tempo_diario_dispositivos_eletronicos'
]

# Filtramos apenas as colunas que existem no df_analise
cols_existentes = [c for c in cols_impacto if c in df_analise.columns]
correlacoes = df_analise[cols_existentes].corr()['Escala_Obesidade'].drop('Escala_Obesidade').sort_values(ascending=False)

# 4. Visualização do Impacto Geral
plt.figure(figsize=(12, 10))
colors = ['#d73027' if x > 0 else '#4575b4' for x in correlacoes]
ax = sns.barplot(x=correlacoes.values, y=correlacoes.index, palette=colors)

plt.title('Impacto na Progressão da Classe de Peso (Correlação Global)')
plt.xlabel('Força do Impacto (Correlação de Pearson)')
plt.ylabel('Variável Comportamental / Perfil')
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Adicionando os valores de correlação nas barras
for i, v in enumerate(correlacoes.values):
    ax.text(v, i, f' {v:.2f}', va='center', fontsize=10, fontweight='bold')

plt.show()
```
