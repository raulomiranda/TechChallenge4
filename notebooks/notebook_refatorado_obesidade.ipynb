# Refatoração do Pipeline de Obesidade: Foco em Negócio e Limpeza de Viés
Neste notebook, abordaremos a predição da classe de obesidade sob uma ótica de negócio. Nosso primeiro passo será demonstrar o vazamento de dados (data leakage) e a falta de variância em algumas features. Em seguida, implementaremos uma seleção de variáveis (Feature Selection) rigorosa para treinar dois modelos: um biométrico e outro estritamente comportamental/genético.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix
from sklearn.preprocessing import LabelEncoder
import joblib
import pickle
import os

import warnings
warnings.filterwarnings('ignore')

# Carregar os dados
dados = pd.read_csv('../dados/Obesity.csv')

# Renomear as colunas
df_dados_ptbr = dados.rename(columns={
    "Gender": "Genero", 
    "Age": "Idade", 
    "Height": "Altura", 
    "Weight": "Peso", 
    "family_history": "Historico_familiar_sobrepeso", 
    "FAVC": "Consumo_alimentos_hipercaloricos", 
    "FCVC": "Frequencia_consumo_vegetais", 
    "NCP": "Numero_refeicoes_diarias", 
    "CAEC": "Consumo_entre_refeicoes", 
    "SMOKE": "Fumante", 
    "CH2O": "Consumo_agua_diario", 
    "SCC": "Monitora_ingestao_calorica", 
    "FAF": "Frequencia_semanal_atividade_fisica", 
    "TUE": "Tempo_diario_dispositivos_eletronicos", 
    "CALC": "Consumo_alcoolico", 
    "MTRANS": "Meio_transporte_principal",
    "Obesity": "Classe_peso_corporal"
})

# Arredondando as colunas de escala para valores inteiros (como feito no notebook original)
colunas_escala = ["Frequencia_consumo_vegetais", "Numero_refeicoes_diarias", "Consumo_agua_diario", "Frequencia_semanal_atividade_fisica", "Tempo_diario_dispositivos_eletronicos"]
for col in colunas_escala:
    df_dados_ptbr[col] = df_dados_ptbr[col].round().astype(int)

df_dados_ptbr.head()
```

## 1. Análise Exploratória de Dados (EDA) - Evidenciando o Viés

```python
## Análise Estatística Percentual por Categoria de Peso
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

features_para_analisar = [
    'Genero', 'Historico_familiar_sobrepeso', 'Consumo_alimentos_hipercaloricos', 
    'Frequencia_consumo_vegetais', 'Numero_refeicoes_diarias', 'Consumo_entre_refeicoes', 
    'Fumante', 'Consumo_agua_diario', 'Monitora_ingestao_calorica', 
    'Frequencia_semanal_atividade_fisica', 'Tempo_diario_dispositivos_eletronicos', 
    'Consumo_alcoolico', 'Meio_transporte_principal'
]

ordem_classes = [
    'Insufficient_Weight', 'Normal_Weight', 'Overweight_Level_I', 
    'Overweight_Level_II', 'Obesity_Type_I', 'Obesity_Type_II', 'Obesity_Type_III'
]

for feature in features_para_analisar:
    # Calculando a tabela de contingência (frequência absoluta)
    df_cont = pd.crosstab(df_dados_ptbr['Classe_peso_corporal'], df_dados_ptbr[feature])
  
    # Normalizando por linha para obter a porcentagem dentro de cada classe de peso
    df_perc = df_cont.div(df_cont.sum(axis=1), axis=0) * 100
  
    # Reorganizando para plotagem
    df_plot = df_perc.reset_index().melt(id_vars='Classe_peso_corporal')
  
    plt.figure(figsize=(14, 7))
    ax = sns.barplot(data=df_plot, x='Classe_peso_corporal', y='value', hue=feature, order=ordem_classes, palette='viridis')
  
    plt.title(f'Distribuição Percentual de {feature} por Classe de Obesidade')
    plt.ylabel('Porcentagem (%)')
    plt.xlabel('Classe de Peso')
    plt.xticks(rotation=45)
    plt.legend(title=feature, bbox_to_anchor=(1.05, 1), loc=2)
    plt.ylim(0, 110)
  
    # Adicionando rótulos de porcentagem nas barras
    for p in ax.patches:
        if p.get_height() > 0:
            ax.annotate(f'{p.get_height():.1f}%', 
                        (p.get_x() + p.get_width() / 2., p.get_height()), 
                        ha = 'center', va = 'center', 
                        xytext = (0, 9), 
                        textcoords = 'offset points', 
                        fontsize=8, rotation=90)
  
    plt.tight_layout()
    plt.show()
    print('-'*100)
```

### Conclusão da Análise de Viés
Os gráficos revelam um **vazamento de dados grave na variável Gênero**, especialmente nas classes de Obesidade Tipo II (quase 100% masculino na base de dados) e Tipo III (quase 100% feminino). Isso significa que, se utilizarmos o Gênero, o modelo tenderá a apenas decorar essas classes baseado no sexo, falhando em generalizar os hábitos. 

Além disso, notamos uma **falta de variância em variáveis como Fumante e Monitoramento de Calorias**, nas quais a vasta maioria das respostas recai em uma única categoria (ex: 'no'). Sem variância, essas variáveis não provêm ganho de informação para a divisão em uma árvore de decisão, sendo classificadas como ruído.

## 2. Feature Selection e Limpeza

Abaixo aplicamos as seguintes regras:
* **Features de Ouro (Manter):**
  * `Historico_familiar_sobrepeso`: Preditor categórico de alto impacto.
  * `Consumo_alimentos_hipercaloricos`: Progressão comportamental clara.
  * `Frequencia_semanal_atividade_fisica`: Marcador forte de sedentarismo nas classes mais altas.
  * `Idade` (Age): Variável contínua importante para os cortes nas árvores de decisão.
* **Features de Suporte (Manter):**
  * `Frequencia_consumo_vegetais` e `Numero_refeicoes_diarias`: Variáveis comportamentais que ajudam a compor o perfil do paciente.
* **Features de Ruído e Viés (Descartar Imediatamente):**
  * `Genero`: Possui vazamento de dados grave nas classes II e III.
  * `Fumante` e `Monitora_ingestao_calorica`: Quase 100% de respostas em uma única categoria (falta de variância).
  * `Consumo_entre_refeicoes` e `Meio_transporte_principal`: Comportamento engessado e artificial nas classes-alvo.

## 3. Engenharia de Features (Criação do IMC)

```python
# 3. Engenharia de Features (Criação do IMC)
df_dados_ptbr['IMC'] = df_dados_ptbr['Peso'] / (df_dados_ptbr['Altura'] ** 2)

# Filtrar as colunas desejadas para Feature Selection
features_ouro = ['Historico_familiar_sobrepeso', 'Consumo_alimentos_hipercaloricos', 'Frequencia_semanal_atividade_fisica', 'Idade']
features_suporte = ['Frequencia_consumo_vegetais', 'Numero_refeicoes_diarias']

# Mantemos as originais (Peso e Altura) apenas para o IMC, mas elas serão descartadas no treinamento.
df_modelo = df_dados_ptbr[features_ouro + features_suporte + ['IMC', 'Classe_peso_corporal']].copy()

# Tratamento das variaveis categoricas mantidas (Label Encoding)
label_encoders = {}
categorical_cols = ['Historico_familiar_sobrepeso', 'Consumo_alimentos_hipercaloricos']

for col in categorical_cols:
    le = LabelEncoder()
    df_modelo[col] = le.fit_transform(df_modelo[col])
    label_encoders[col] = le
    
# Encoder do Target para garantir a ordenação correta das classes
le_target = LabelEncoder()
df_modelo['Classe_peso_corporal'] = le_target.fit_transform(df_modelo['Classe_peso_corporal'])
label_encoders['Classe_peso_corporal'] = le_target

df_modelo.head()
```

## 4. Estratégia de Dupla Modelagem (O Dilema do IMC)

Para demonstrar valor de negócio e evitar que o algoritmo funcione apenas como uma "calculadora de IMC glorificada", treinaremos e avaliaremos **dois modelos Random Forest** separados:

* **Modelo 1 (A Linha de Base Biométrica):**
  * **Features de treinamento:** Todas as variáveis mantidas + `IMC` (remova `Peso` e `Altura` originais).
  * **Objetivo:** Demonstrar a altíssima acurácia esperada (próxima a 99%) quando o algoritmo tem acesso aos dados biométricos diretos.
* **Modelo 2 (A Abordagem Comportamental/Genética):**
  * **Features de treinamento:** Apenas as Features de Ouro e de Suporte listadas no Passo 2 (remova `Peso`, `Altura` e também remova o `IMC`).
  * **Objetivo:** Responder à pergunta de negócio: "É possível prever a classe de obesidade apenas pelos hábitos e genética do paciente, sem colocá-lo na balança?".
  * **Requisito:** Este modelo precisa alcançar uma assertividade (acurácia) acima de 75%.

```python
# Preparacao dos Dados
X_bio = df_modelo[features_ouro + features_suporte + ['IMC']]
X_comp = df_modelo[features_ouro + features_suporte]
y = df_modelo['Classe_peso_corporal']

# Divisao Treino e Teste (Stratified para manter proporcao de classes)
X_train_bio, X_test_bio, y_train, y_test = train_test_split(X_bio, y, test_size=0.3, random_state=42, stratify=y)
X_train_comp, X_test_comp, _, _ = train_test_split(X_comp, y, test_size=0.3, random_state=42, stratify=y)

# Treinamento do Modelo 1 (Biométrico)
rf_bio = RandomForestClassifier(random_state=42, n_estimators=200, max_depth=10)
rf_bio.fit(X_train_bio, y_train)

# Treinamento do Modelo 2 (Comportamental)
rf_comp = RandomForestClassifier(random_state=42, n_estimators=200, max_depth=15)
rf_comp.fit(X_train_comp, y_train)
```

## 5. Avaliação e Exportação

Para **ambos os modelos**, vamos exibir:
* Classification Report completo (Precision, Recall, F1-Score).
* Matriz de Confusão visual (usando Seaborn/Matplotlib).
* Gráfico de Feature Importance.

```python
def plot_evaluation(model, X_test, y_test, model_name, feature_names):
    # Predicoes
    y_pred = model.predict(X_test)
    
    # 1. Classification Report
    print(f"==================== {model_name} ====================")
    print("Classification Report:")
    target_names = le_target.inverse_transform(np.unique(y_test))
    print(classification_report(y_test, y_pred, target_names=target_names))
    
    # 2. Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)
    plt.figure(figsize=(10, 7))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=target_names, yticklabels=target_names)
    plt.title(f'Matriz de Confusão - {model_name}')
    plt.ylabel('Classe Real')
    plt.xlabel('Classe Prevista')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
    
    # 3. Feature Importance
    importances = model.feature_importances_
    idx = np.argsort(importances)[::-1]
    
    plt.figure(figsize=(10, 5))
    sns.barplot(x=importances[idx], y=np.array(feature_names)[idx], palette='viridis')
    plt.title(f'Feature Importance - {model_name}')
    plt.xlabel('Importância Relativa')
    plt.ylabel('Features')
    plt.tight_layout()
    plt.show()
    print("-" * 80)

# Avaliacao Modelo Biometrico
plot_evaluation(rf_bio, X_test_bio, y_test, "Modelo 1 (A Linha de Base Biométrica)", X_bio.columns)

# Avaliacao Modelo Comportamental
plot_evaluation(rf_comp, X_test_comp, y_test, "Modelo 2 (A Abordagem Comportamental/Genética)", X_comp.columns)
```

### Exportação do Modelo
Ao final, exportamos os modelos nos formatos solicitados (`modelo_biometrico.pkl` e `modelo_comportamental.pkl`), utilizando a biblioteca `pickle`.

```python
import os

os.makedirs('../modelos', exist_ok=True)

# Exportando o Modelo 1 (Biometrico)
with open('../modelos/modelo_biometrico.pkl', 'wb') as f:
    pickle.dump(rf_bio, f)

# Exportando o Modelo 2 (Comportamental)
with open('../modelos/modelo_comportamental.pkl', 'wb') as f:
    pickle.dump(rf_comp, f)

# Exportando os Encoders e Informações Relevantes adicionais usando pickle
with open('../modelos/label_encoders_rf.pkl', 'wb') as f:
    pickle.dump(label_encoders, f)
    
with open('../modelos/features_comportamental_rf.pkl', 'wb') as f:
    pickle.dump(list(X_comp.columns), f)

print("Ambos os modelos ('modelo_biometrico.pkl' e 'modelo_comportamental.pkl') foram exportados com sucesso na pasta '../modelos/'!")
```
