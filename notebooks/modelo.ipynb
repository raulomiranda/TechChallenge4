**Preparando para o modelo**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler, OneHotEncoder, OrdinalEncoder, LabelEncoder
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Modelos para avaliação
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, ExtraTreesClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

try:
    from imblearn.over_sampling import SMOTE
    _SMOTE_AVAILABLE = True
except ImportError:
    _SMOTE_AVAILABLE = False
```

### Definição dos Transformadores

```python
class GroupObesityClasses(BaseEstimator, TransformerMixin):
    def fit(self, df, y=None): return self
    def transform(self, df):
        df = df.copy()
        def agrupar(classe):
            if classe in ["Insufficient_Weight", "Normal_Weight", "Overweight_Level_I", "Overweight_Level_II"]:
                return "Não obeso"
            elif classe == "Obesity_Type_I": return "Obeso tipo 1"
            elif classe == "Obesity_Type_II": return "Obeso tipo 2"
            elif classe == "Obesity_Type_III": return "Obeso tipo 3"
            return classe
        if 'Classe_peso_corporal' in df.columns: df['Classe_peso_corporal'] = df['Classe_peso_corporal'].apply(agrupar)
        return df

class DropFeatures(BaseEstimator, TransformerMixin):
    def __init__(self, drop=['Peso', 'Altura', 'Genero']): self.drop = drop
    def fit(self, df, y=None): return self
    def transform(self, df):
        df = df.copy()
        cols = [f for f in self.drop if f in df.columns]
        return df.drop(cols, axis=1)

class OneHotEncodingNames(BaseEstimator, TransformerMixin):
    def __init__(self, cols=['Meio_transporte_principal']): 
        self.cols = cols
        self.enc = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
    def fit(self, df, y=None): 
        self.enc.fit(df[self.cols])
        return self
    def transform(self, df):
        df = df.copy()
        oh = pd.DataFrame(self.enc.transform(df[self.cols]), columns=self.enc.get_feature_names_out(self.cols), index=df.index)
        return pd.concat([oh, df.drop(self.cols, axis=1)], axis=1)

class OrdinalFeature(BaseEstimator, TransformerMixin):
    def __init__(self):
        self.map_bin = {"F": 1, "M": 0}
        self.map_ord = {"no": 0, "Sometimes": 1, "Frequently": 2, "Always": 3}
    def fit(self, df, y=None): return self
    def transform(self, df):
        df = df.copy()
        if 'Genero' in df.columns: df['Genero'] = df['Genero'].map(self.map_bin)
        for c in ['Consumo_entre_refeicoes', 'Consumo_alcoolico']:
            if c in df.columns: df[c] = df[c].map(self.map_ord)
        return df

class MinMax(BaseEstimator, TransformerMixin):
    def __init__(self, cols=None):
        self.cols = cols or ['Idade', 'Frequencia_consumo_vegetais', 'Numero_refeicoes_diarias', 'Consumo_agua_diario', 'Frequencia_semanal_atividade_fisica', 'Tempo_diario_dispositivos_eletronicos']
        self.scaler = MinMaxScaler()
    def fit(self, df, y=None): 
        self.scaler.fit(df[self.cols])
        return self
    def transform(self, df):
        df = df.copy()
        df[self.cols] = self.scaler.transform(df[self.cols])
        return df

class Oversample(BaseEstimator, TransformerMixin):
    def __init__(self, seed): self.seed = seed
    def fit(self, df, y=None): return self
    def transform(self, df):
        if 'Classe_peso_corporal' in df.columns and _SMOTE_AVAILABLE:
            sm = SMOTE(sampling_strategy='not majority', random_state=self.seed)
            X_bal, y_bal = sm.fit_resample(df.drop('Classe_peso_corporal', axis=1), df['Classe_peso_corporal'])
            return pd.concat([pd.DataFrame(X_bal), pd.DataFrame(y_bal, columns=['Classe_peso_corporal'])], axis=1)
        return df
```

#**Avaliação de Estabilidade Multi-Seed**

```python
SEEDS = [10, 4232, 9999]
model_list = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier(n_estimators=100),
    "Extra Trees": ExtraTreesClassifier(n_estimators=100),
    "AdaBoost": AdaBoostClassifier(),
    "Gradient Boosting": GradientBoostingClassifier(),
    "SVM": SVC(probability=True)
}

final_results = pd.DataFrame(index=model_list.keys())

print("Iniciando bateria de testes...")
for s in SEEDS:
    acc_per_seed = []
    # Split
    train_df, test_df = train_test_split(df_dados_ptbr, test_size=0.2, random_state=s, stratify=df_dados_ptbr['Classe_peso_corporal'])
    
    # Pipeline
    pipeline = Pipeline([
        ('group', GroupObesityClasses()),
        ('drop', DropFeatures()),
        ('ordinal', OrdinalFeature()),
        ('onehot', OneHotEncodingNames()),
        ('scale', MinMax())
    ])
    
    df_tr = pipeline.fit_transform(train_df)
    val_pipe = Pipeline(pipeline.steps[:-1])
    df_te = val_pipe.transform(test_df)
    
    X_tr, y_tr = df_tr.drop('Classe_peso_corporal', axis=1), df_tr['Classe_peso_corporal']
    X_te, y_te = df_te.drop('Classe_peso_corporal', axis=1), df_te['Classe_peso_corporal']
    
    for name, model in model_list.items():
        model.set_params(random_state=s) if hasattr(model, 'random_state') else None
        model.fit(X_tr, y_tr)
        acc = accuracy_score(y_te, model.predict(X_te))
        acc_per_seed.append(acc)
    
    final_results[f'Seed_{s}'] = acc_per_seed

final_results['Média'] = final_results.mean(axis=1)
final_results = final_results.sort_values(by='Média', ascending=False)

print("\n=== RANKING DE ESTABILIDADE ===")
display(final_results)

melhor_nome = final_results.index[0]
print(f"\nO modelo campeão (mais estável) é: {melhor_nome}")
```

### Treinamento Final e Visualização

```python
# Treinamos o vencedor com o SEED principal (4232) para exportação
MAIN_SEED = 4232
train_df, test_df = train_test_split(df_dados_ptbr, test_size=0.2, random_state=MAIN_SEED, stratify=df_dados_ptbr['Classe_peso_corporal'])

final_pipeline = Pipeline([
    ('group', GroupObesityClasses()),
    ('drop', DropFeatures()),
    ('ordinal', OrdinalFeature()),
    ('onehot', OneHotEncodingNames()),
    ('scale', MinMax())
])

df_tr_final = final_pipeline.fit_transform(train_df)
val_pipeline = Pipeline(final_pipeline.steps[:-1])
df_te_final = val_pipeline.transform(test_df)

X_tr_final, y_tr_final = df_tr_final.drop('Classe_peso_corporal', axis=1), df_tr_final['Classe_peso_corporal']
X_te_final, y_te_final = df_te_final.drop('Classe_peso_corporal', axis=1), df_te_final['Classe_peso_corporal']

vencedor = model_list[melhor_nome]
vencedor.set_params(random_state=MAIN_SEED) if hasattr(vencedor, 'random_state') else None
vencedor.fit(X_tr_final, y_tr_final)

y_pred = vencedor.predict(X_te_final)
labels = ["Não obeso", "Obeso tipo 1", "Obeso tipo 2", "Obeso tipo 3"]

print(f"\n=== RELATÓRIO FINAL: {melhor_nome} ===")
print(classification_report(y_te_final, y_pred, target_names=labels))

cm = confusion_matrix(y_te_final, y_pred)
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.title(f'Matriz de Confusão Final ({melhor_nome})')
plt.show()

# Gráfico de Importância das Features
if hasattr(vencedor, 'feature_importances_'):
    importancias = vencedor.feature_importances_
    features = X_tr_final.columns
    df_importancia = pd.DataFrame({'Feature': features, 'Importancia': importancias}).sort_values(by='Importancia', ascending=False)

    plt.figure(figsize=(10, 8))
    sns.barplot(x='Importancia', y='Feature', data=df_importancia, palette='viridis')
    plt.title(f'Importância das Features - {melhor_nome}')
    plt.xlabel('Impacto no Resultado (Importância)')
    plt.ylabel('Feature')
    plt.show()
else:
    print(f"O modelo {melhor_nome} não suporta visualização direta de importância de features (coeficientes).")
```

### Exportação para Joblib

```python
final_data = {
    'preprocessor': val_pipeline,
    'model': vencedor,
    'features': X_tr_final.columns.tolist(),
    'target_names': labels,
    'model_name': melhor_nome
}

joblib.dump(final_data, 'modelo_obesidade.joblib')
print("Arquivo 'modelo_obesidade.joblib' gerado com sucesso!")
```
